## 2. El contrato exacto

El archivo `api_sesion28.h` declara la interfaz que el harness sabe probar.
Tu implementación **debe** respetar esas firmas; si las cambias, el harness no compila.

```c
#include "api_sesion28.h"

void plegar(char *dst, int cap, const char *res, const char *op, const char *a, const char *b);
```

- Solo pliega cuando ambos operandos son enteros completos; nada de prefijos ni texto sobrante.
- x + 0 y x * 1 se reducen a x; x * 0 se reduce a 0.
- Si el divisor es cero, no pliegues: deja la instrucción original y no provoques una excepción.

> El harness compila con `-Wall -Wextra -Werror`: una advertencia del compilador cuenta como error.
>
> El harness compila `harness_sesion28_c.c` junto con `artesano_sesion28_c.c`

````python
%%writefile artesano_sesion28_c.c
/* artesano_sesion28_c.c - Fase 1: El Artesano (sin IA)
 * Plegado de constantes e identidades.
 *
 * Implementa lo que declara api_sesion28.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion28.h"
#include <limits.h>
#include <stdio.h>
#include <string.h>

static int hay(const char *s) {
    return s != NULL && s[0] != '\0';
}

/* Entero completo: '-' opcional y solo digitos, nada de espacios, '+', 0x ni texto sobrante.
 * Ademas tiene que caber en un int, que es el tamano del destino. */
static int entero(const char *s, long long *v) {
    const char *p = s;
    int neg = (*p == '-');
    if (neg) p++;
    if (*p == '\0') return 0;                        /* agregado en la auditoria: "-" solo no es un numero */
    if (strlen(p) > 10) return 0;                    /* un int tiene a lo mas 10 digitos */
    long long x = 0;
    for (; *p; p++) {
        if (*p < '0' || *p > '9') return 0;
        x = x * 10 + (*p - '0');
    }
    x = neg ? -x : x;
    if (x < INT_MIN || x > INT_MAX) return 0;
    *v = x;
    return 1;
}

/* Calcula en long long y solo pliega si el resultado cabe en un int.
 * Asi INT_MAX + 1 o INT_MIN / -1 se quedan como estaban, igual que en ejecucion. */
static int calcular(char o, long long x, long long y, long long *r) {
    switch (o) {
    case '+': *r = x + y; break;
    case '-': *r = x - y; break;
    case '*': *r = x * y; break;
    case '/':
        if (y == 0) return 0;                        /* entre cero no se pliega */
        *r = x / y;
        break;
    default: return 0;
    }
    return *r >= INT_MIN && *r <= INT_MAX;
}

void plegar(char *dst, int cap, const char *res, const char *op, const char *a, const char *b) {
    if (dst == NULL || cap <= 0) return;
    dst[0] = '\0';
    if (!hay(res) || !hay(op) || !hay(a) || !hay(b)) return;

    long long x = 0, y = 0, r = 0;
    int ca = entero(a, &x), cb = entero(b, &y);
    char o = (strlen(op) == 1) ? op[0] : '?';
    int w;

    /* 1) dos literales: un solo literal */
    if (ca && cb && calcular(o, x, y, &r)) {
        w = snprintf(dst, (size_t)cap, "%s = %lld", res, r);
    } else {
        /* 2) identidades: el operando que sobrevive, o 0 */
        const char *queda = NULL;
        if (o == '+' && cb && y == 0) queda = a;                     /* x + 0 */
        else if (o == '+' && ca && x == 0) queda = b;                /* 0 + x */
        else if (o == '-' && cb && y == 0) queda = a;                /* x - 0 (0 - x no) */
        else if ((o == '*' || o == '/') && cb && y == 1) queda = a;  /* x * 1, x / 1 */
        else if (o == '*' && ca && x == 1) queda = b;                /* 1 * x */
        else if (o == '*' && ((cb && y == 0) || (ca && x == 0))) queda = "0";
        /* 0 / x no se toca: si x vale 0 en ejecucion, el programa original falla */

        w = queda ? snprintf(dst, (size_t)cap, "%s = %s", res, queda)
                  : snprintf(dst, (size_t)cap, "%s = %s %s %s", res, a, op, b);   /* 3) la original */
    }
    if (w < 0 || w >= cap) dst[0] = '\0';            /* cortado no: mejor vacio */
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle
la misma función a la IA, y pega el resultado en la celda de `ia_sesion28_c.c`. La
firma debe ser idéntica a la del contrato; si la IA la cambia, corrígela antes de
compilar.


````python
%%writefile ia_sesion28_c.c
/* ia_sesion28_c.c - Fase 2: El Cientifico
 *
 * Pega aqui el codigo que te devolvio la IA, con la MISMA firma
 * que declara api_sesion28.h. No puede traer main().
 */
#include "api_sesion28.h"
#include <errno.h>
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

/* Devuelve 1 si s es un literal entero completo y guarda su valor. */
static int es_entero(const char *s, long *valor) {
    if (s == NULL || *s == '\0') return 0;
    char *fin = NULL;
    errno = 0;
    long v = strtol(s, &fin, 10);
    if (errno == ERANGE || fin == s || *fin != '\0') return 0;  /* texto sobrante o fuera de rango */
    *valor = v;
    return 1;
}

/* Escribe en dst sin truncar: si no cabe completo, deja la cadena vacia. */
static void escribir(char *dst, int cap, int escritos) {
    if (escritos < 0 || escritos >= cap) dst[0] = '\0';
}

void plegar(char *dst, int cap, const char *res, const char *op, const char *a, const char *b) {
    if (dst == NULL || cap <= 0) return;
    dst[0] = '\0';
    if (res == NULL || op == NULL || a == NULL || b == NULL) return;
    if (*res == '\0' || *op == '\0' || *a == '\0' || *b == '\0') return;

    long va = 0, vb = 0;
    int a_const = es_entero(a, &va);
    int b_const = es_entero(b, &vb);
    char o = (op[1] == '\0') ? op[0] : '\0';

    /* Plegado de constantes: los dos operandos son literales. */
    if (a_const && b_const && o != '\0') {
        long r = 0;
        int plegable = 1;
        switch (o) {
            case '+': r = va + vb; break;
            case '-': r = va - vb; break;
            case '*': r = va * vb; break;
            case '/':
                if (vb == 0) plegable = 0;   /* division entre cero: no se pliega */
                else r = va / vb;
                break;
            default: plegable = 0; break;
        }
        if (plegable) {
            escribir(dst, cap, snprintf(dst, (size_t)cap, "%s = %ld", res, r));
            return;
        }
    }

    /* Identidades algebraicas. */
    if (o == '+' && b_const && vb == 0) {               /* x + 0 -> x */
        escribir(dst, cap, snprintf(dst, (size_t)cap, "%s = %s", res, a));
        return;
    }
    if (o == '*' && b_const && vb == 1) {               /* x * 1 -> x */
        escribir(dst, cap, snprintf(dst, (size_t)cap, "%s = %s", res, a));
        return;
    }
    if (o == '*' && b_const && vb == 0) {               /* x * 0 -> 0 */
        escribir(dst, cap, snprintf(dst, (size_t)cap, "%s = 0", res));
        return;
    }

    /* Sin simplificacion posible: la instruccion original. */
    escribir(dst, cap, snprintf(dst, (size_t)cap, "%s = %s %s %s", res, a, op, b));
}
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion28_c.c
/* harness_sesion28_c.c - auditoria automatica de la Sesion 28
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion28_c.c artesano_sesion28_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion28.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 28 - Plegado de constantes e identidades ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { char b[64] = ""; plegar(b, 64, "_t0", "+", "3", "5"); verificar_str("3 + 5", "_t0 = 8", b); }
    { char b[64] = ""; plegar(b, 64, "_t0", "*", "2", "3"); verificar_str("2 * 3", "_t0 = 6", b); }
    { char b[64] = ""; plegar(b, 64, "_t0", "-", "7", "10"); verificar_str("7 - 10", "_t0 = -3", b); }
    { char b[64] = ""; plegar(b, 64, "_t0", "+", "x", "0"); verificar_str("x + 0", "_t0 = x", b); }
    { char b[64] = ""; plegar(b, 64, "_t0", "+", "x", "y"); verificar_str("x + y (sin plegar)", "_t0 = x + y", b); }
    { char b[64] = ""; plegar(b, 64, "_t0", "/", "10", "0"); verificar_str("10 / 0 (no plegar)", "_t0 = 10 / 0", b); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion28.md" "\n");
    return 0;
}
````
