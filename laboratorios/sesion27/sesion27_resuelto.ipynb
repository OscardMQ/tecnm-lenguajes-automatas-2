## 1. ¿Qué vas a construir?

Para un arreglo 1D emites la multiplicación del índice por el ancho y la lectura base[_tK]. Para uno 2D primero aplanas i * filas + j, luego multiplicas por el ancho y lees. Los temporales se numeran desde *temporales.

Al final de la sesión tendrás: **Dos funciones en C que emiten el cálculo de offset y la lectura indexada para arreglos 1D y 2D.**

## 3. Cómo se califica

### Casos que evalúa el harness

| # | Caso | Resultado esperado |
| :-: | :--- | :--- |
| 1 | `1D: i * 4` | `_t0 = i * 4` |
| 2 | `1D: A[_t0]` | `_t1 = A[_t0]` |
| 3 | `2D: i * 3` | `_t0 = i * 3` |
| 4 | `2D: offset por ancho` | `_t2 = _t1 * 4` |
| 5 | `2D: M[_t2]` | `_t3 = M[_t2]` |
| 6 | `base NULL (sin acceso)` | 0 |

````python
%%writefile artesano_sesion27_c.c
/* artesano_sesion27_c.c - Fase 1: El Artesano (sin IA)
 * Desplazamientos para arreglos 1D y 2D.
 *
 * Implementa lo que declara api_sesion27.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion27.h"
#include <limits.h>
#include <stdio.h>
#include <string.h>

/* "" cuenta igual que NULL (lo de la sesion 26) */
static int hay(const char *s) {
    return s != NULL && s[0] != '\0';
}

/* Todo o nada: copia las k lineas solo si todas cupieron completas y hay lugar.
 * Cada linea crea un temporal nuevo, asi que el contador avanza k. */
static int volcar(char lineas[][64], int max, char tmp[][64], const int *w, int k, int *temporales) {
    for (int i = 0; i < k; i++)
        if (w[i] < 0 || w[i] >= 64) return 0;
    if (max < k) return 0;
    for (int i = 0; i < k; i++) memcpy(lineas[i], tmp[i], 64);
    *temporales += k;
    return k;
}

/* _tK = indice * ancho   y   _tK+1 = base[_tK]. Regresa cuantas lineas escribio. */
int emitir_acceso(char lineas[][64], int max, const char *base, const char *indice, int ancho, int *temporales) {
    if (lineas == NULL || temporales == NULL || !hay(base) || !hay(indice)) return 0;
    if (ancho <= 0) return 0;                        /* agregado en la auditoria: con ancho 0 todo i lee A[0] */
    int t = *temporales;
    if (t < 0 || t > INT_MAX - 2) return 0;          /* ni _t-1 ni un contador que se desborde */

    char tmp[2][64];
    int w[2];
    w[0] = snprintf(tmp[0], 64, "_t%d = %s * %d", t, indice, ancho);     /* el offset siempre lleva el ancho */
    w[1] = snprintf(tmp[1], 64, "_t%d = %s[_t%d]", t + 1, base, t);
    return volcar(lineas, max, tmp, w, 2, temporales);
}

/* Aplanado: i * filas + j, despues por el ancho y la lectura.
 * filas es lo largo de cada fila (en A[N][M] es la M). */
int emitir_acceso_2d(char lineas[][64], int max, const char *base, const char *i, const char *j, int filas, int ancho, int *temporales) {
    if (lineas == NULL || temporales == NULL || !hay(base) || !hay(i) || !hay(j)) return 0;
    if (filas <= 0 || ancho <= 0) return 0;          /* agregado en la auditoria */
    int t = *temporales;
    if (t < 0 || t > INT_MAX - 4) return 0;

    char tmp[4][64];
    int w[4];
    w[0] = snprintf(tmp[0], 64, "_t%d = %s * %d", t, i, filas);
    w[1] = snprintf(tmp[1], 64, "_t%d = _t%d + %s", t + 1, t, j);
    w[2] = snprintf(tmp[2], 64, "_t%d = _t%d * %d", t + 2, t + 1, ancho);
    w[3] = snprintf(tmp[3], 64, "_t%d = %s[_t%d]", t + 3, base, t + 2);
    return volcar(lineas, max, tmp, w, 4, temporales);
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle

````python
%%writefile ia_sesion27_c.c
/* ia_sesion27_c.c - Fase 2: El Cientifico
 *
 * Pega aqui el codigo que te devolvio la IA, con la MISMA firma
 * que declara api_sesion27.h. No puede traer main().
 */
#include "api_sesion27.h"
#include <stdio.h>
#include <string.h>

#define ANCHO_LINEA 64
#define MAX_LINEAS_ACCESO 4

/* Un texto es valido solo si no es NULL ni la cadena vacia. */
static int texto_valido(const char *s) {
    return s != NULL && *s != '\0';
}

/* Copia las lineas preparadas al destino solo si todas estan completas
 * y hay espacio para todas (todo o nada). Devuelve cuantas copio. */
static int confirmar(char lineas[][64], int max, char buffer[][ANCHO_LINEA],
                     const int *escritos, int cantidad) {
    if (max < cantidad) return 0;
    for (int k = 0; k < cantidad; k++) {
        if (escritos[k] < 0 || escritos[k] >= ANCHO_LINEA) return 0;  /* linea truncada */
    }
    for (int k = 0; k < cantidad; k++) {
        memcpy(lineas[k], buffer[k], ANCHO_LINEA);
    }
    return cantidad;
}

int emitir_acceso(char lineas[][64], int max, const char *base, const char *indice,
                  int ancho, int *temporales) {
    if (lineas == NULL || temporales == NULL) return 0;
    if (!texto_valido(base) || !texto_valido(indice) || ancho <= 0) return 0;

    char buffer[MAX_LINEAS_ACCESO][ANCHO_LINEA];
    int escritos[2];
    int t = *temporales;

    /* Offset: el indice siempre se multiplica por el ancho del elemento. */
    escritos[0] = snprintf(buffer[0], ANCHO_LINEA, "_t%d = %s * %d", t, indice, ancho);
    /* Lectura indexada desde la base con el offset calculado. */
    escritos[1] = snprintf(buffer[1], ANCHO_LINEA, "_t%d = %s[_t%d]", t + 1, base, t);

    int copiadas = confirmar(lineas, max, buffer, escritos, 2);
    if (copiadas > 0) *temporales = t + 2;
    return copiadas;
}

int emitir_acceso_2d(char lineas[][64], int max, const char *base, const char *i,
                     const char *j, int filas, int ancho, int *temporales) {
    if (lineas == NULL || temporales == NULL) return 0;
    if (!texto_valido(base) || !texto_valido(i) || !texto_valido(j)) return 0;
    if (filas <= 0 || ancho <= 0) return 0;

    char buffer[MAX_LINEAS_ACCESO][ANCHO_LINEA];
    int escritos[4];
    int t = *temporales;

    /* Aplanado en orden de filas: i * filas + j. */
    escritos[0] = snprintf(buffer[0], ANCHO_LINEA, "_t%d = %s * %d", t, i, filas);
    escritos[1] = snprintf(buffer[1], ANCHO_LINEA, "_t%d = _t%d + %s", t + 1, t, j);
    /* Escalado por el ancho del elemento y lectura indexada. */
    escritos[2] = snprintf(buffer[2], ANCHO_LINEA, "_t%d = _t%d * %d", t + 2, t + 1, ancho);
    escritos[3] = snprintf(buffer[3], ANCHO_LINEA, "_t%d = %s[_t%d]", t + 3, base, t + 2);

    int copiadas = confirmar(lineas, max, buffer, escritos, 4);
    if (copiadas > 0) *temporales = t + 4;
    return copiadas;
}
````

````python
%%writefile harness_sesion27_c.c
/* harness_sesion27_c.c - auditoria automatica de la Sesion 27
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion27_c.c artesano_sesion27_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion27.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 27 - Desplazamientos para arreglos 1D y 2D ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { char l[8][64] = {{0}}; int t = 0; emitir_acceso(l, 8, "A", "i", 4, &t); verificar_str("1D: i * 4", "_t0 = i * 4", l[0]); }
    { char l[8][64] = {{0}}; int t = 0; emitir_acceso(l, 8, "A", "i", 4, &t); verificar_str("1D: A[_t0]", "_t1 = A[_t0]", l[1]); }
    { char l[16][64] = {{0}}; int t = 0; emitir_acceso_2d(l, 16, "M", "i", "j", 3, 4, &t); verificar_str("2D: i * 3", "_t0 = i * 3", l[0]); }
    { char l[16][64] = {{0}}; int t = 0; emitir_acceso_2d(l, 16, "M", "i", "j", 3, 4, &t); verificar_str_largo("2D: offset por ancho", "_t2 = _t1 * 4", l[2]); }
    { char l[16][64] = {{0}}; int t = 0; emitir_acceso_2d(l, 16, "M", "i", "j", 3, 4, &t); verificar_str("2D: M[_t2]", "_t3 = M[_t2]", l[3]); }
    { char l[8][64] = {{0}}; int t = 0; int k = emitir_acceso(l, 8, NULL, "i", 4, &t); verificar_int("base NULL (sin acceso)", (long)(0), (long)(k)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion27.md" "\n");
    return 0;
}
````
